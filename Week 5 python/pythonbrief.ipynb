Préparation et nettoyage des données

```python
#Remplissage des valeurs manquantes
df['director']
df['director']= df['director'].fillna('Unknown')
df.head()
```

```python
df['cast'] = df['cast'].fillna('Unknown')
df.head()
```

```python
#Vérification des doublons
df.duplicated().sum()
```

```python
df.isnull().sum()
```

```python
list(df['rating'].unique())
```

```python
#Valeurs trouvés dans la colonne rating qui appartiennet à la colonne duration
values = ['74 min','84 min','66 min']
mask = df['rating'].isin(values)
df.loc[mask,'duration'] = df.loc[mask,'rating']
df.loc[mask,'rating']= np.nan
```

```python
list(df['rating'].unique())
```

```python
#Remplir les rating manquants
most_frequent_rating=df['rating'].mode()[0]
print(most_frequent_rating)
df['rating'] = df['rating'].fillna(most_frequent_rating)
```

```python
#Suppression les dqtes nulls 
df = df.dropna(subset=['date_added'])
```

```python
#Conversion en datetime
df['date_added'] = df['date_added'].str.strip()
df['date_added'] = pd.to_datetime(df['date_added'])
df['date_added']
```

```python
#Création des colonnes dérivées de l'année d'ajout
df['year_added'] = df['date_added'].dt.year
df['month_added'] = df['date_added'].dt.month
df['day_added'] = df['date_added'].dt.day
df.head()
```

```python
#Séparer la colonne duration en deux colonnes ( durée en nombres , unités pour chaque type ( movie/tv show))
df[['duration_nbr','duration_unit']] = df['duration'].str.extract(r'(\d+)\s*(\w+)')
df = df.drop('duration', axis=1)
df.head()
```

```python
#Remplacer les valeurs nulls dans country par top 5 pays fréquents
country_values = list(df['country'].value_counts().head(5).index)
df['country']=df['country'].fillna(np.random.choice(country_values))
```

 Visualisations et analyse métier

```python
df['listed_in'] = df['listed_in'].str.split(', ')
df = df.explode('listed_in')

df['country'] = df['country'].str.split(', ')
df = df.explode('country')
```

```python
#Répartition par classification d’âge (rating)
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(10,6))
sns.countplot(data=df, y='rating', order=df['rating'].value_counts().index)
plt.title("Distribution des contenus par classification d'âge")
plt.xlabel("Nombre de contenus")
plt.ylabel("Classification")
plt.show()
```

```python
#Proportion de films vs séries (type)
plt.figure(figsize=(6,6))
df['type'].value_counts().plot.pie(autopct='%1.1f%%')
plt.title("Proportion Films vs Séries")
plt.ylabel("")
plt.show()
```

```python
# Nombre de contenus ajoutés chaque année (year_added)
plt.figure(figsize=(10,6))
df['year_added'].value_counts().sort_index().plot(kind='line')
plt.title("Nombre de contenus ajoutés par année")
plt.xlabel("Année")
plt.ylabel("Nombre de contenus")
plt.show()
```

```python
# Évolution du catalogue : films vs séries par année
plt.figure(figsize=(10,6))
sns.countplot(data=df, x='year_added', hue='type')
plt.xticks(rotation=45)
plt.title("Évolution Films vs Séries par année")
plt.show()
```

```python
# Nombre de séries de 2011 à 2021
series_year = df[(df['type']=='TV Show') & 
                 (df['year_added'].between(2011,2021))]

plt.figure(figsize=(8,5))
series_year['year_added'].value_counts().sort_index().plot(kind='bar')
plt.title("Nombre de séries ajoutées par année (2011-2021)")
plt.xlabel("Année")
plt.ylabel("Nombre de séries")
plt.show()
```

```python
# Nombre de productions par pays
df['country'] = df['country'].str.split(',')

df_exploded = df.explode('country')

df_exploded['country'] = df_exploded['country'].str.strip()

top_countries = df_exploded['country'].value_counts().head(10)

plt.figure(figsize=(10,6))
sns.barplot(x=top_countries.values, y=top_countries.index)
plt.title('Top 10 pays producteurs')
plt.xlabel('Nombre de productions')
plt.ylabel('Pays')
plt.show()
```

```python
# Nombre de productions par pays et par classification d’âges
df['country'] = df['country'].str.split(',')
df_exploded = df.explode('country')
df_exploded['country'] = df_exploded['country'].str.strip()

pays = df_exploded['country'].value_counts().head(10)

plt.figure()
sns.countplot(data=pays, x= 'country', hue='rating')
plt.title('Nombre de productions par pays')
plt.xlabel('Pays')
plt.ylabel('Nombre de production')
plt.show()
```
